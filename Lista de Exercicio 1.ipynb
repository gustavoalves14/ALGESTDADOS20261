```python
# 20 Simulador de Pedágio Inteligente

print('# Simulador de Pedágio Inteligente')
tarifa_moto = 5.00
tarifa_carro = 10.00
tarifa_caminhao = 20.00
tarifa_onibus = 15.00

while True:
    tipo_veiculo = input('Informe o tipo de veículo (moto, carro, caminhao, onibus): ').lower()
    match tipo_veiculo:
        case 'moto':
            tarifa_base = tarifa_moto
            break
        case 'carro':
            tarifa_base = tarifa_carro
            break
        case 'caminhao':
            tarifa_base = tarifa_caminhao
            break
        case 'onibus':
            tarifa_base = tarifa_onibus
            break
        case _:
            print('Tipo de veículo inválido. Por favor, tente novamente.')

possui_tag = input('Possui tag de pagamento automático (s/n)? ').lower()

while True:
    try:
        hora_atual = int(input('Informe a hora atual (0-23): '))
        if 0 <= hora_atual <= 23:
            break
        else:
            print('Hora inválida. Digite um número entre 0 e 23.')
    except ValueError:
        print('Entrada inválida. Digite um número inteiro para a hora.')

tarifa_final = tarifa_base
percentual_desconto = 0
percentual_acrescimo = 0

if possui_tag == 's':
    desconto = tarifa_final * 0.10
    tarifa_final -= desconto
    percentual_desconto = 10
    print(f'Desconto de 10% aplicado pela tag automática: -R$ {desconto:.2f}')

if (7 <= hora_atual <= 9) or (17 <= hora_atual <= 19):
    acrescimo = tarifa_final * 0.20
    tarifa_final += acrescimo
    percentual_acrescimo = 20
    print(f'Acréscimo de 20% aplicado por horário de pico: +R$ {acrescimo:.2f}')

print('\n--- Detalhes da Tarifa ---')
print(f'Tipo de veículo: {tipo_veiculo.capitalize()}')
print(f'Tarifa base: R$ {tarifa_base:.2f}')
if percentual_desconto > 0:
    print(f'Desconto da tag: {percentual_desconto}%')
if percentual_acrescimo > 0:
    print(f'Acréscimo por horário de pico: {percentual_acrescimo}%')
print(f'Tarifa final a pagar: R$ {tarifa_final:.2f}')
```
