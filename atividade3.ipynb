```python
#02
#Senha Segura
#Defina uma senha correta no código.
#Usando while, peça a senha ao usuário repetidamente até que ele acerte. A cada tentativa errada, exiba 'Senha incorreta, tente novamente.'. Ao acertar, exiba 'Acesso liberado!' e quantas tentativas foram necessárias.
contador = 0
senha = "spice grils"
while True:
    senha_digitada = input("Digite a senha:")
    if senha_digitada == senha:
        print(f"Acesso liberado! Usou {contador} tentativas")
        break
    else:
        print("Senha incorreta, tente novamente.")
        contador += 1
```
