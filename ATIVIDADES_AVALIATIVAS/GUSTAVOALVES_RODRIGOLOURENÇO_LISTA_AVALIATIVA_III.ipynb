```
#Questão 1
#Crie uma lista com 15 números inteiros digitados pelo usuário. Em seguida, remova todos os números repetidos, mantendo a ordem original.

numeros = []
for i in range(15):
    n = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(n)

print("Lista original:", numeros)

lista_sem_repetidos = []
for n in numeros:
    if n not in lista_sem_repetidos:
        lista_sem_repetidos.append(n)

print("Lista sem repetidos:", lista_sem_repetidos)
```

```
#Questão 2
#Escreva um programa que leia 10 nomes e 10 idades. Armazene os nomes em uma lista e as idades em outra. Em seguida, exiba o nome da pessoa mais velha.

nomes = []
idades = []

for i in range(10):
    nome = input(f"Digite o {i+1}º nome: ")
    idade = int(input(f"Digite a idade de {nome}: "))
    nomes.append(nome)
    idades.append(idade)

mais_velho_nome = nomes[0]
mais_velho_idade = idades[0]

for i in range(1, 10):
    if idades[i] > mais_velho_idade:
        mais_velho_idade = idades[i]
        mais_velho_nome = nomes[i]

print("A pessoa mais velha é:", mais_velho_nome)
```

```
#Questão 3
#Faça um programa que leia uma lista de 10 números e divida todos os valores ímpares por 3 e os pares por 2, substituindo os valores originais. (Use estruturas condicionais.)

numeros = []

for i in range(10):
    n = int(input(f"Digite o {i+1}º número: "))
    numeros.append(n)

for i in range(10):
    if numeros[i] % 2 == 0:
        numeros[i] = numeros[i] / 2
    else:
        numeros[i] = numeros[i] / 3

print("Lista modificada:", numeros)
```

```
#Questão 4
#Solicite ao usuário uma frase. Armazene em uma lista cada palavra que começa com uma vogal. Exiba a lista resultante.

frase = input("Digite uma frase: ")

resultado = []
palavra = ""

for caractere in frase + " ":
    if caractere != " ":
        palavra += caractere
    else:
        if palavra != "":
            primeira_letra = palavra[0]
            if primeira_letra in "aeiouAEIOU":
                resultado.append(palavra)
            palavra = ""

print("Palavras que começam com vogal:", resultado)
```

```
#Questão 5
#Leia 10 números inteiros e armazene-os em uma lista. Em seguida, crie uma segunda lista apenas com os números primos da primeira.

numeros = []
for i in range(10):
    n = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(n)

primos = []
for n in numeros:
    if n > 1:
        eh_primo = True
        for i in range(2, int(n**0.5) + 1):
            if n % i == 0:
                eh_primo = False
                break
        if eh_primo:
            primos.append(n)

print("Lista original:", numeros)
print("Lista de primos:", primos)
```

```
#Questão 7
#Crie duas listas com 10 números inteiros aleatórios entre 1 e 100. Gere uma terceira lista contendo apenas os elementos que estão presentes em ambas (interseção via estruturas de repetição).

lista1 = []
lista2 = []

print("Digite 10 números para a primeira lista:")
for i in range(10):
    n = int(input(f"{i+1}º número: "))
    lista1.append(n)

print("\nDigite 10 números para a segunda lista:")
for i in range(10):
    n = int(input(f"{i+1}º número: "))
    lista2.append(n)

intersecao = []
for n in lista1:
    if n in lista2 and n not in intersecao:
        intersecao.append(n)

print("\nLista 1:", lista1)
print("Lista 2:", lista2)
print("Interseção:", intersecao)
```

```
#Questão 8
#Faça um programa que simule o sorteio de 6 números da Mega-Sena (entre 1 e 60) sem repetição. Armazene em uma lista e exiba.

numeros = []

print("Digite 6 números distintos entre 1 e 60:")

while len(numeros) < 6:
    n = int(input(f"Número {len(numeros)+1}: "))
    if 1 <= n <= 60:
        if n not in numeros:
            numeros.append(n)
        else:
            print("Número repetido! Digite outro.")
    else:
        print("Número fora do intervalo! Digite entre 1 e 60.")

print("Números sorteados:", numeros)
```

```
#Questão 12
#Gere uma matriz 5×5 com números aleatórios entre 1 e 99. Calcule a média dos elementos de cada linha e armazene os resultados em uma nova lista.

matriz = []
for linha in range(5):
    nova_linha = []
    for coluna in range(5):
        valor = ((linha + 2) * (coluna + 5) * 3) % 99 + 1
        nova_linha.append(valor)
    matriz.append(nova_linha)

print("Matriz 5x5:")
for linha in matriz:
    print(linha)

medias = []
for linha in matriz:
    soma = 0
    for elemento in linha:
        soma += elemento
    media = soma / len(linha)
    medias.append(media)

print("\nMédias das linhas:")
print(medias)

```

```
#Questão 14
#Faça um programa que leia uma matriz 3×3 e verifique se ela é simétrica (igual à sua transposta). Use estruturas de repetição e condicionais.

matriz = []
for botafogo in range(3):
    linha = [int(input("Digite um número: ")) for botafogo in range(3)]
    matriz.append(linha)

print("\nMatriz 3x3:")
for linha in matriz:
    print(linha)

simetrica = 1
for linha in range(3):
    for coluna in range(3):
        if matriz[linha][coluna] != matriz[coluna][linha]:
            simetrica = 0

if simetrica == 1:
    print("\nA matriz é simétrica.")
else:
    print("\nA matriz NÃO é simétrica.")
```

```
#Questão 16
# Crie uma matriz 4×4 com números fornecidos pelo usuário. Encontre o menor valor de cada coluna e armazene em uma lista separada.

matriz = []

for i in range(4):
    linha = []

    for j in range(4):
        valor = int(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(valor)

    matriz.append(linha)

menores = []

for j in range(4):
    menor = matriz[0][j]

    for i in range(4):
        if matriz[i][j] < menor:
            menor = matriz[i][j]

    menores.append(menor)

print("\nMatriz:")
for linha in matriz:
    print(linha)

print("\nMenor valor de cada coluna:")
print(menores)
```

```
#Questão 17
# Dada uma matriz 3×3 com valores reais, calcule o determinante (sem usar bibliotecas externas). Use apenas listas e estruturas de repetição.

matriz = []

for i in range(3):
    linha = []

    for j in range(3):
        valor = float(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(valor)

    matriz.append(linha)

determinante = (
    matriz[0][0] * (matriz[1][1] * matriz[2][2] - matriz[1][2] * matriz[2][1])
    - matriz[0][1] * (matriz[1][0] * matriz[2][2] - matriz[1][2] * matriz[2][0])
    + matriz[0][2] * (matriz[1][0] * matriz[2][1] - matriz[1][1] * matriz[2][0])
)

print("Determinante =", determinante)
```

```
#Questão 18
# Faça um programa que leia duas matrizes 3×3 e calcule a matriz soma e a matriz diferença, armazenando os resultados em novas listas.

matriz1 = []

for i in range(3):
    linha = []
    for j in range(3):
        linha.append(int(input(f"Matriz 1 [{i}][{j}]: ")))
    matriz1.append(linha)

matriz2 = []

for i in range(3):
    linha = []
    for j in range(3):
        linha.append(int(input(f"Matriz 2 [{i}][{j}]: ")))
    matriz2.append(linha)

soma = []
diferenca = []

for i in range(3):
    linha_soma = []
    linha_diferenca = []

    for j in range(3):
        linha_soma.append(matriz1[i][j] + matriz2[i][j])
        linha_diferenca.append(matriz1[i][j] - matriz2[i][j])

    soma.append(linha_soma)
    diferenca.append(linha_diferenca)

print("\nMatriz Soma:")
for linha in soma:
    print(linha)

print("\nMatriz Diferença:")
for linha in diferenca:
    print(linha)
```

```
#Questão 19
# Solicite dois conjuntos de 10 elementos fornecidos pelo usuário. Mostre a interseção, união e diferença (A − B e B − A).

A = set()
B = set()

print("Conjunto A")

for i in range(10):
    numero = int(input(f"Digite o {i+1}º elemento: "))
    A.add(numero)

print("\nConjunto B")

for i in range(10):
    numero = int(input(f"Digite o {i+1}º elemento: "))
    B.add(numero)

print("\nInterseção:", A & B)
print("União:", A | B)
print("Diferença A - B:", A - B)
print("Diferença B - A:", B - A)
```

```
#Questão 25
# Crie um dicionário que armazene o nome e o salário de 5 funcionários. Calcule o reajuste de 10% para salários inferiores a R$ 2.000 e atualize o dicionário. (Use estruturas condicionais.)

funcionarios = {}

for i in range(5):
    nome = input(f"Digite o nome do {i+1}º funcionário: ")
    salario = float(input(f"Digite o salário de {nome}: R$ "))

    funcionarios[nome] = salario

for nome in funcionarios:
    if funcionarios[nome] < 2000:
        funcionarios[nome] *= 1.10
print("\nSalários atualizados:")

for nome, salario in funcionarios.items():
    print(f"{nome}: R$ {salario:.2f}")
```

```
#Questão 26
# Faça um programa que armazene em um dicionário o nome de 5 alunos e uma lista com suas 3 notas. Depois, exiba o nome e a média de cada aluno.

alunos = {}

for i in range(5):
    nome = input(f"Digite o nome do {i+1}º aluno: ")

    notas = []

    for j in range(3):
        nota = float(input(f"Digite a {j+1}ª nota de {nome}: "))
        notas.append(nota)

    alunos[nome] = notas

print("\nMédias dos alunos:")

for nome, notas in alunos.items():
    media = sum(notas) / len(notas)
    print(f"{nome}: Média = {media:.2f}")
```

```
#Questão 28
# Solicite ao usuário 10 palavras e armazene em um dicionário onde a chave é a primeira letra e o valor é uma lista de palavras que começam com essa letra.

palavras_letra = {}

for i in range(10):
    palavra = input(f"Digite a {i+1}ª palavra: ").lower()

    primeira_letra = palavra[0]

    if primeira_letra not in palavras_letra:
        palavras_letra[primeira_letra] = []

    palavras_letra[primeira_letra].append(palavra)

print("\nPalavras agrupadas pela primeira letra:")

for letra, lista_palavras in palavras_letra.items():
    print(f"{letra}: {lista_palavras}")
```

```
#Questão 29
# Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes. Encontre o produto com maior variação de preço.

produtos = {}

for i in range(5):
    nome = input(f"Digite o nome do {i+1}º produto: ")

    precos = []

    for j in range(3):
        preco = float(input(f"Preço na loja {j+1}: R$ "))
        precos.append(preco)

    produtos[nome] = precos

maior_variacao = 0
produto_maior_variacao = ""

for nome, precos in produtos.items():
    variacao = max(precos) - min(precos)

    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior_variacao = nome

print("\nProduto com maior variação de preço:")
print(produto_maior_variacao)
print(f"Variação: R$ {maior_variacao:.2f}")
```

```
#Questão 30
# Implemente um programa que leia uma frase e armazene em um dicionário o número de ocorrências de cada palavra (ignorando maiúsculas/minúsculas). (Use estruturas de repetição.)

frase = input("Digite uma frase: ").lower()

palavras = frase.split()

ocorrencias = {}

for palavra in palavras:
    if palavra in ocorrencias:
        ocorrencias[palavra] += 1
    else:
        ocorrencias[palavra] = 1

print("\nQuantidade de ocorrências de cada palavra:")

for palavra, quantidade in ocorrencias.items():
    print(f"{palavra}: {quantidade}")
```
