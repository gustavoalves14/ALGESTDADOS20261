```python
# Questão 1

# Uma empresa de médio porte deseja realizar um levantamento sobre a remuneração de seus funcionários.
# Desenvolva um programa que leia os salários de 10 funcionários, um por vez, e, ao final, informe a média salarial da empresa, o maior salário registrado e quantos funcionários recebem acima de R$ 3.000.
# Os valores informados são válidos e positivos.

soma_salarios = 0
maior_salario = 0
acima_3000 = 0
quantidade_funcionarios = 10
valor_invalido = False

for i in range(1, quantidade_funcionarios + 1):
    salario = float(input(f"Digite o salário do funcionário {i}: "))

    if salario < 0:
        print("Erro: Salário não pode ser negativo. O programa será cancelado.")
        valor_invalido = True
        break

    soma_salarios += salario
    if salario > maior_salario:
        maior_salario = salario
    if salario > 3000:
        acima_3000 += 1

if valor_invalido:
    print("Processamento cancelado devido a um salário negativo.")
else:
    media = soma_salarios / quantidade_funcionarios

    print("Média salarial da empresa:", media)
    print("Maior salário registrado:", maior_salario)
    print("Quantidade de funcionários com salário acima de R$ 3.000:", acima_3000)
```

```python
# Questão 2

# Em uma instituição de ensino, um professor deseja analisar o desempenho de uma turma sem saber previamente a quantidade de alunos.
# Desenvolva um programa que leia as notas dos alunos até que seja digitado o valor -1 (fim da entrada).
# Ao final, calcule a média da turma, informe quantos alunos foram aprovados (nota ≥ 7) e quantos foram reprovados.

soma_notas = 0
quantidade_alunos = 0
aprovados = 0
reprovados = 0
nota_invalida = False

while True:
    nota = float(input("Digite a nota do aluno (ou -1 para encerrar): "))

    if nota == -1:
        break

    if nota < 0 and nota != -1:
        print("Erro: Nota inválida!")
        nota_invalida = True
        break

    soma_notas += nota
    quantidade_alunos += 1
    if nota >= 7:
        aprovados += 1
    elif nota < 7:
        reprovados += 1
if nota_invalida:
    print("Erro no cidigo pela nota ser invalida.")
elif quantidade_alunos > 0:
    media = soma_notas / quantidade_alunos
    print(f"Média da turma: {media:.2f}")
    print("Quantidade de alunos aprovados:", aprovados)
    print("Quantidade de alunos reprovados:", reprovados)
else:
    print("Nenhuma nota válida foi informada.")
```

```python
# Questão 3

# Um sistema bancário simples precisa simular operações de saque em uma conta corrente.
# O programa deve iniciar com um saldo informado pelo usuário e permitir saques sucessivos até que seja digitado 0.
# A cada tentativa, verifique se há saldo suficiente, atualize o saldo quando possível e informe ao usuário quando o valor exceder o saldo disponível.

invalid_input_detected = False

saldo_inicial = float(input("Digite o saldo inicial da conta: "))

if saldo_inicial < 0:
    print("Erro: Saldo inicial não pode ser negativo. O programa será cancelado.")
    invalid_input_detected = True
else:
    saldo = saldo_inicial
    while True:
        saque = float(input("Digite o valor do saque (ou 0 para encerrar): "))

        if saque == 0:
            break

        if saque < 0:
            print("Erro: O valor do saque não pode ser negativo. O programa será cancelado.")
            invalid_input_detected = True
            break

        if saque <= saldo:
            saldo -= saque
            print(f"Saque realizado. Saldo atual: R$ {saldo:.2f}")
        else:
            print("Saldo insuficiente para realizar o saque.")

if invalid_input_detected:
    print("Operações bancárias canceladas devido a um valor inválido.")
else:
    print(f"Operações encerradas. Saldo final: R$ {saldo:.2f}")
```

```python
# Questão 4

# Uma pesquisa demográfica está sendo realizada para identificar o perfil etário de um grupo de pessoas.
# O programa deve ler as idades informadas até que o usuário digite 0, encerrando a coleta.
# Ao final, apresente a média das idades, a maior idade registrada e quantas pessoas possuem mais de 18 anos.

soma_idades = 0
quantidade = 0
maior_idade = 0
maiores_de_18 = 0
invalid_input_detected = False

while True:
    idade = int(input("Digite a idade (ou 0 para encerrar): "))
    if idade == 0:
        break

    if idade < 0:
        print("Erro: Idade não pode ser negativa. O programa será cancelado.")
        invalid_input_detected = True
        break

    soma_idades += idade
    quantidade += 1
    if idade > maior_idade:
        maior_idade = idade
    if idade > 18:
        maiores_de_18 += 1

if invalid_input_detected:
    print("Processamento cancelado devido a uma idade negativa.")
elif quantidade > 0:
    media = soma_idades / quantidade
    print("Média das idades:", media)
    print("Maior idade registrada:", maior_idade)
    print("Quantidade de pessoas com mais de 18 anos:", maiores_de_18)
else:
    print("Nenhuma idade foi informada.")
```

```python
# Questão 5

# Um sistema de segurança controla o acesso a um ambiente restrito.
# O programa deve solicitar uma senha numérica e permitir até três tentativas.
# Caso a senha correta (2024) seja informada dentro desse limite, o acesso deve ser liberado.
# Caso contrário, após três tentativas incorretas, o sistema deve bloquear o acesso e informar essa condição ao usuário.

tentativas = 0
senha_correta = 2024
acesso_liberado = False

while tentativas < 3:
    senha = int(input("Digite a senha: "))
    if senha == senha_correta:
        acesso_liberado = True
        break
    else:
        tentativas += 1

if acesso_liberado:
    print("Acesso liberado.")
else:
    print("Acesso bloqueado após 3 tentativas incorretas.")
```

```python
# Questão 7

# Um sistema de análise numérica deve processar uma sequência de valores informados pelo usuário.
# O programa deve ler números inteiros até que seja digitado 0.
# Durante a execução, conte quantos números são pares, quantos são ímpares e calcule a soma de todos os números positivos informados.

pares = 0
impares = 0
soma_positivos = 0

while True:
    numero = int(input("Digite um número inteiro (ou 0 para encerrar): "))
    if numero == 0:
        break
    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1
    if numero > 0:
        soma_positivos += numero

print(f"Quantidade de números pares: {pares}")
print(f"Quantidade de números ímpares: {impares}")
print(f"Soma dos números positivos: {soma_positivos}")
```

```python
# Questão 9

# Um sistema de votação eletrônica simplificado será utilizado em uma eleição interna.
# Os votos devem ser informados por meio de códigos numéricos (1, 2 ou 3), sendo que o valor 0 indica o encerramento da votação.
# O programa deve contabilizar os votos de cada candidato e, ao final, informar qual candidato obteve a maior quantidade de votos.

votos1 = 0
votos2 = 0
votos3 = 0

while True:
    voto = int(input("Digite seu voto (1, 2, 3) ou 0 para encerrar: "))
    if voto == 0:
        break
    elif voto == 1:
        votos1 += 1
    elif voto == 2:
        votos2 += 1
    elif voto == 3:
        votos3 += 1
    else:
        print("Voto inválido.")

if votos1 > votos2 and votos1 > votos3:
    vencedor = 1
elif votos2 > votos1 and votos2 > votos3:
    vencedor = 2
elif votos3 > votos1 and votos3 > votos2:
    vencedor = 3
else:
    vencedor = None

print(f"Votos candidato 1: {votos1}")
print(f"Votos candidato 2: {votos2}")
print(f"Votos candidato 3: {votos3}")

if vencedor:
    print(f"O candidato vencedor foi o {vencedor}.")
else:
    print("Houve empate na votação.")
```

```python
# Questão 11

# Um sistema matemático deve analisar propriedades de números inteiros informados pelo usuário.
# O programa deve ler valores até que seja digitado -1 e, ao final, informar quantos números são múltiplos de 3 e calcular a média apenas desses números.

multiplos_3 = 0
soma_multiplos = 0

while True:
    numero = int(input("Digite um número inteiro (ou -1 para encerrar): "))
    if numero == -1:
        break
    if numero % 3 == 0:
        multiplos_3 += 1
        soma_multiplos += numero

if multiplos_3 > 0:
    media = soma_multiplos / multiplos_3
else:
    media = 0

print(f"Quantidade de números múltiplos de 3: {multiplos_3}")
print(f"Média dos números múltiplos de 3: {media:.2f}")
```

```python
# Questão 14

# Um programa deve analisar um número inteiro informado pelo usuário.
# O sistema deve verificar se o número é primo, apresentar todos os seus divisores e informar a quantidade total de divisores encontrados.

numero = int(input("Digite um número inteiro positivo: "))
invalid_input_detected = False

if numero <= 0:
    print("Erro: O número deve ser inteiro e positivo. O programa será cancelado.")
    invalid_input_detected = True
else:
    divisores = []
    contador = 1

    while contador <= numero:
        if numero % contador == 0:
            divisores.append(contador)
        contador += 1

    quantidade_divisores = 0
    for d in divisores:
        quantidade_divisores += 1

    if quantidade_divisores == 2:
        print(numero, "é primo.")
    else:
        print(numero, "não é primo.")

    print("Divisores de", numero, ":", divisores)
    print("Quantidade total de divisores:", quantidade_divisores)

if invalid_input_detected:
    print("Processamento cancelado devido a um número inválido.")
```

```python
# Questão 20

# Um sistema acadêmico deve processar notas de alunos.
# O programa deve ler notas até que seja digitado -1. Ao final, deve calcular a média das notas,
# informar a maior nota e indicar quantas notas estão abaixo da média.

notas = []
soma = 0
quantidade = 0
invalid_input_detected = False

while True:
    nota = float(input("Digite a nota (ou -1 para encerrar): "))
    if nota == -1:
        break

    if nota < 0: # Note: This check allows -1 to terminate, but flags other negative numbers as invalid.
        print("Erro: A nota não pode ser negativa (exceto -1 para encerrar). O programa será cancelado.")
        invalid_input_detected = True
        break

    notas.append(nota)
    soma += nota
    quantidade += 1

if invalid_input_detected:
    print("Processamento cancelado devido a uma nota negativa inválida.")
elif quantidade > 0:
    media = soma / quantidade

    maior = notas[0]
    for n in notas:
        if n > maior:
            maior = n

    abaixo_media = 0
    for n in notas:
        if n < media:
            abaixo_media += 1

    print(f"Média das notas: {media:.2f}")
    print(f"Maior nota: {maior:.2f}")
    print(f"Quantidade de notas abaixo da média: {abaixo_media}")
else:
    print("Nenhuma nota foi informada.")
```

```python
# Questão 21

# Uma pesquisa eleitoral está sendo realizada para identificar a preferência dos eleitores.
# O programa deve ler votos até que seja digitado 0. Votos válidos são representados por códigos específicos; quaisquer outros valores devem ser considerados inválidos.
# Ao final, informe a quantidade de votos válidos, inválidos e o candidato vencedor.

votos_validos = {1: 0, 2: 0, 3: 0}
invalidos = 0

while True:
    voto = int(input("Digite seu voto (1, 2, 3) ou 0 para encerrar: "))
    if voto == 0:
        break
    if voto in votos_validos:
        votos_validos[voto] += 1
    else:
        invalidos += 1

total_validos = 0
for candidato in votos_validos:
    total_validos += votos_validos[candidato]

if total_validos > 0:
    vencedor = None
    maior_votos = -1
    for candidato in votos_validos:
        if votos_validos[candidato] > maior_votos:
            maior_votos = votos_validos[candidato]
            vencedor = candidato

    print("Quantidade de votos válidos:", total_validos)
    print("Quantidade de votos inválidos:", invalidos)
    print("Candidato vencedor:", vencedor)
else:
    print("Nenhum voto válido foi registrado.")
```

```python
# Questão 22

# Um estacionamento deseja automatizar o cálculo de sua arrecadação diária.
# O programa deve ler o tempo de permanência de cada veículo até que seja digitado 0.
# O valor a ser pago deve ser calculado com base no tempo, com aplicação de desconto para permanências superiores a 5 horas.
# Ao final, apresente o total arrecadado.

total = 0
cancelled_to_negative_input = False

while True:
    horas = float(input("Digite o tempo de permanência do veículo (em horas, ou 0 para encerrar): "))

    if horas == 0:
        break

    if horas < 0:
        print("Erro: O tempo de permanência não pode ser negativo.")
        cancelled_to_negative_input = True
        break

    if horas > 5:
        valor = horas * 8 * 0.90
        valor = horas * 8

    total += valor

if cancelled_to_negative_input:
    print("Processamento cancelado devido a um valor de tempo negativo.")
else:
    print(f"Total arrecadado: R$ {total:.2f}")
```

```python
# Questão 23

# Um sistema deve analisar uma sequência de números inteiros informados pelo usuário até que seja digitado 0.
# Ao final, informe quantos números são divisíveis simultaneamente por 2 e por 3, além de calcular a média de todos os números informados.

soma = 0
quantidade = 0
divisiveis = 0

while True:
    numero = int(input("Digite um número inteiro (ou 0 para encerrar): "))
    if numero == 0:
        break
    soma += numero
    quantidade += 1
    if numero % 2 == 0 and numero % 3 == 0:
        divisiveis += 1

if quantidade > 0:
    media = soma / quantidade
    print("Quantidade de números divisíveis por 2 e 3:", divisiveis)
    print("Média dos números informados:", media)
else:
    print("Nenhum número foi informado.")
```

```python
# Questão 24

# Um sistema escolar precisa classificar o desempenho dos alunos com base em suas médias finais.
# O programa deve ler 8 valores e classificar cada um como aprovado, em recuperação ou reprovado.
# Ao final, apresente a quantidade de alunos em cada categoria.

aprovados = 0
recuperacao = 0
reprovados = 0
odd_number_error = False
negative_grade_error = False

for i in range(8):
    media = float(input(f"Digite a média final do aluno {i+1}: "))

    if media < 0:
        print(f"Erro: A média {media} não pode ser negativa. O programa será cancelado.")
        negative_grade_error = True
        break

    if media == int(media) and int(media) % 2 != 0:
        print(f"Erro: A média {media} é um número ímpar e não é permitida. O programa será cancelado.")
        odd_number_error = True
        break

    if media >= 7:
        aprovados += 1
    elif media >= 5:
        recuperacao += 1
    else:
        reprovados += 1

if odd_number_error or negative_grade_error:
    print("Processamento cancelado devido a uma média inválida.")
else:
    print("Quantidade de alunos aprovados:", aprovados)
    print("Quantidade de alunos em recuperação:", recuperacao)
    print("Quantidade de alunos reprovados:", reprovados)
```

```python
# Questão 25

# Uma empresa deseja analisar o valor de suas vendas.
# O programa deve ler valores até que seja digitado 0, calcular a média das vendas e informar quantas vendas possuem valor superior à média calculada.

vendas = []
soma = 0
quantidade = 0
invalid_input_detected = False

while True:
    valor = float(input("Digite o valor da venda (ou 0 para encerrar): "))
    if valor == 0:
        break

    if valor < 0:
        print("Erro: O valor da venda não pode ser negativo. O programa será cancelado.")
        invalid_input_detected = True
        break

    vendas.append(valor)
    soma += valor
    quantidade += 1

if invalid_input_detected:
    print("Processamento cancelado devido a um valor de venda negativo.")
elif quantidade > 0:
    media = soma / quantidade

    acima_media = 0
    for v in vendas:
        if v > media:
            acima_media += 1

    print("Média das vendas:", media)
    print("Quantidade de vendas acima da média:", acima_media)
else:
    print("Nenhuma venda foi informada.")
```

```python
# Questão 26

# Um sistema deve coletar a idade de diversas pessoas até que seja digitado 0.
# Ao final, deve calcular a média das idades e informar quantas pessoas são maiores de idade (acima de 18 anos).

soma = 0
quantidade = 0
maiores_idade = 0
invalid_input_detected = False

while True:
    idade = int(input("Digite a idade (ou 0 para encerrar): "))
    if idade == 0:
        break

    if idade < 0:
        print("Erro: Idade não pode ser negativa. O programa será cancelado.")
        invalid_input_detected = True
        break

    soma += idade
    quantidade += 1
    if idade > 18:
        maiores_idade += 1

if invalid_input_detected:
    print("Processamento cancelado devido a uma idade negativa.")
elif quantidade > 0:
    media = soma / quantidade
    print("Média das idades:", media)
    print("Quantidade de pessoas maiores de idade:", maiores_idade)
else:
    print("Nenhuma idade foi informada.")
```

```python
# Questão 27

# Um programa deve ler um número inteiro N e, utilizando repetição, exibir todos os números de 1 até N.
# Para cada número exibido, o sistema deve informar se ele é par ou ímpar.

N = int(input("Digite um número inteiro: "))

if N < 1: # N should be positive for the range 1 to N
    print("Erro: O número N deve ser um inteiro positivo. O programa será cancelado.")
else:
    for i in range(1, N + 1):
        if i % 2 == 0:
            print(i, "é par")
        else:
            print(i, "é ímpar")
```

```python
# Questão 28

# Um sistema deve analisar números inteiros informados pelo usuário até que seja digitado -1.
# Ao final, informe quantos números são primos e calcule a soma dos números que não são primos.

soma_nao_primos = 0
quantidade_primos = 0

while True:
    numero = int(input("Digite um número inteiro (ou -1 para encerrar): "))
    if numero == -1:
        break

    if numero >= 2:
        primo = True
        divisor = 2
        while divisor * divisor <= numero:
            if numero % divisor == 0:
                primo = False
                break
            divisor += 1
    else:
        primo = False

    if primo:
        quantidade_primos += 1
    else:
        soma_nao_primos += numero

print("Quantidade de números primos:", quantidade_primos)
print("Soma dos números não primos:", soma_nao_primos)
```

```python
# Questão 29

# Um sistema de avaliação deve ler notas até que seja digitado 0, calcular a média geral e informar quantas notas estão acima da média.

soma = 0
quantidade = 0
notas = []
invalid_input_detected = False

while True:
    nota = float(input("Digite a nota (ou 0 para encerrar): "))
    if nota == 0:
        break

    if nota < 0: # Note: This check allows 0 to terminate, but flags other negative numbers as invalid.
        print("Erro: A nota não pode ser negativa (exceto 0 para encerrar). O programa será cancelado.")
        invalid_input_detected = True
        break

    soma += nota
    quantidade += 1
    notas.append(nota)

if invalid_input_detected:
    print("Processamento cancelado devido a uma nota negativa inválida.")
elif quantidade > 0:
    media = soma / quantidade
    acima_media = 0
    for n in notas:
        if n > media:
            acima_media += 1

    print("Média geral:", media)
    print("Quantidade de notas acima da média:", acima_media)
else:
    print("Nenhuma nota foi informada.")
```
