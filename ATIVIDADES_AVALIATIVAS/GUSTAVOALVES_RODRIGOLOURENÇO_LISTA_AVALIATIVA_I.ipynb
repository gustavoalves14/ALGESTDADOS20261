```
#Questão 1

#Leia três números inteiros. Sem usar listas, determine e exiba qual é o maior, qual é o menor e qual é o do meio.
#Use apenas variáveis, operadores relacionais e condicionais aninhadas. Considere também o caso em que dois ou mais números sejam iguais.

a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))
c = int(input("Digite o terceiro número: "))

if a >= b and a >= c:
    maior = a
elif b >= a and b >= c:
    maior = b
else:
    maior = c

if a <= b and a <= c:
    menor = a
elif b <= a and b <= c:
    menor = b
else:
    menor = c

if (a != maior and a != menor) or (a == b == c):
    meio = a
elif (b != maior and b != menor):
    meio = b
else:
    meio = c

print("Maior:", maior)
print("Menor:", menor)
print("Do meio:", meio)
```

```
#Questão 9

#Leia uma temperatura e a unidade de origem usando match/case ("C" para Celsius, "F" para Fahrenheit ou "K" para Kelvin).
#Valide que temperaturas em Kelvin não sejam negativas. Calcule e exiba a temperatura convertida nas outras duas escalas.
#Use variáveis descritivas como temp_celsius, temp_fahrenheit e temp_kelvin. Use case _ para unidade inválida.

temperatura = float(input("Digite a temperatura: "))
unidade = input("Digite a unidade de origem (C/F/K): ")

match unidade:
    case "C":
        temp_celsius = temperatura
        temp_fahrenheit = (temp_celsius * 9/5) + 32
        temp_kelvin = temp_celsius + 273.15
        print(f"Celsius: {temp_celsius:}")
        print(f"Fahrenheit: {temp_fahrenheit:}")
        print(f"Kelvin: {temp_kelvin:}")

    case "F":
        temp_fahrenheit = temperatura
        temp_celsius = (temp_fahrenheit - 32) * 5/9
        temp_kelvin = temp_celsius + 273.15
        print(f"Fahrenheit: {temp_fahrenheit:}")
        print(f"Celsius: {temp_celsius:}")
        print(f"Kelvin: {temp_kelvin:}")

    case "K":
        temp_kelvin = temperatura
        if temp_kelvin < 0:
            print("Temperatura inválida em Kelvin (não pode ser negativa).")
        else:
            temp_celsius = temp_kelvin - 273.15
            temp_fahrenheit = (temp_celsius * 9/5) + 32
            print(f"Kelvin: {temp_kelvin:}")
            print(f"Celsius: {temp_celsius:}")
            print(f"Fahrenheit: {temp_fahrenheit:}")

    case _:
        print("Unidade inválida.")
```

```
#Questão 15

#Leia a hora atual (0-23) e o tipo de usuário usando match/case ("admin", "operador", "visitante"). Valide a hora.
#Para cada tipo, use if aninhado com elif para verificar o turno (manhã: 6-11, tarde: 12-17, noite: 18-23, madrugada: 0-5).
#Regras: admin acessa sempre, operador apenas manhã e tarde, visitante apenas tarde. Exiba se o acesso é permitido ou negado e o turno atual.

hora = int(input("Digite a hora atual (0-23): "))
usuario = input("Digite o tipo de usuário (admin, operador, visitante): ")

if hora < 0 or hora > 23:
    print("Hora inválida.")
else:
    if (hora >= 6 and hora <= 11):
        turno = "Manhã"
    elif (hora >= 12 and hora <= 17):
        turno = "Tarde"
    elif (hora >= 18 and hora <= 23):
        turno = "Noite"
    else:
        turno = "Madrugada"

    match usuario:
        case "admin":
            print(f"Acesso permitido para {usuario}. Turno atual: {turno}.")
        case "operador":
            if (turno == "Manhã") or (turno == "Tarde"):
                print(f"Acesso permitido para {usuario}. Turno atual: {turno}.")
            else:
                print(f"Acesso negado para {usuario}. Turno atual: {turno}.")
        case "visitante":
            if (turno == "Tarde"):
                print(f"Acesso permitido para {usuario}. Turno atual: {turno}.")
            else:
                print(f"Acesso negado para {usuario}. Turno atual: {turno}.")
        case _:
            print("Tipo de usuário inválido.")
```
