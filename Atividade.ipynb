```python
# Leia um número de 1 a 7 e exiba o nome do dia correspondente (1 = Segunda, ..., 7 = Domingo). Para qualquer outro número, exiba "Número inválido".
dia = int(input("Dia da semana:"))
match dia:
    case 1 :print("Domingo")
    case 2 :print("Segunda-Feira")
    case 3 :print("Terça feira")
    case 4 :print("Quarta-Feira")
    case 5 :print("Quinta-Feira")
    case 6 :print("Sexta-Feira")
    case 7 :print("Sabado")
    case _:print("Inexistente")
```

```python
# Leia uma nota. Valide se está entre 0 e 10 usando encadeamento de comparações (0 <= nota <= 10). Se inválida, exiba "Nota inválida!". Se válida, exiba "Nota registrada: [nota]".
nota = float(input("Digite uma nota: "))
if 0 <= nota <= 10:
    print(f"Nota registrada {nota}")
    print("Nota válida")
else:
    print("Nota inválida!")
```

```python
# Leia a idade. Classifique: menor de 12 → "Criança", 12 a 17 → "Adolescente", 18 a 59 → "Adulto", 60 ou mais → "Idoso". Valide que a idade não é negativa.
idade = int(input("Digite sua idade: "))
if idade < 0 or idade > 120:
    print("Idade inválida")
elif 0 < idade < 12:
    print("Criança")
    print("Idade valida")
elif 12 <= idade < 18:
    print("Adolescente")
    print("Idade valida")
elif 18 <= idade < 60:
    print("Adulto")
    print("Idade valida")
elif 60 <= idade:
    print("Idoso")
    print("Idade valida")
```
